# Load Data
Data ist stored as an Excel file. So, we use Pandas to load data

```python
def load_data(file_path):
    data= pd.read_excel(
        file_path,
        engine='openpyxl',
        sheet_name=0
        )

    # komplett leere Spalten entfernen
    data = data.dropna(axis=1, how="all")
     # komplett leere Zeilen entfernen
    data = data.dropna(axis=0, how="all")

    return data

file_path="../Data/useCarDaten.xlsx"
import os
print("is path avaiable:",os.path.exists("../Data/useCarDaten.xlsx"))
data= load_data(file_path)
print(data.shape)
#print(data.head())
data.head(40)
```

First of all, we split test data for later analysis

```python

x=data.drop(['Sale price'], axis=1)
y=data['Sale price']

x_train, x_test, y_train, y_test= train_test_split(x, y, test_size=0.2, random_state=42)
data= x_train.copy()
data['Sale price']= y_train
print(data.isnull().sum())
#print(x.info())
```

### Remove all missing value

```python
# Remove all rows that have any missing values
data.dropna(inplace=True)

# Optional: reset index after dropping
data.reset_index(drop=True, inplace=True)

print(data.isnull().sum())
```

## Handle Duplicate Rows

```python
duplicate_rows=data.duplicated()

if duplicate_rows.any():
    print(" There ist Duplicate rows")
    print(f"Number of duplicate rows: {duplicate_rows.sum()}")
    data= data.drop_duplicates()
else:
    print("No duplicate rows")
print(f"Shape of data after removing duplicates: {data.shape}")
```
