C() tells the model that this feature is categorical (even if ordinal).

**ANOVA** is used to compare the means of three or more groups to see if at least one group mean is significantly different from the others.

One-way ANOVA: One independent categorical variable.

Two-way ANOVA: Two independent categorical variables

#### ANOVA Test

```python
# ANOVA test
for col in categorical_cols:
    if data[col].nunique() < 2:  # Skip columns with only 1 unique value
        print(f'{col}: skipped (only 1 unique value)')
        continue
    
    formula = f'Q("Sale price") ~ C(Q("{col}"))' #define the formular
    model = ols(formula, data=data).fit() # fit the model
    anova_table = sm.stats.anova_lm(model, typ=1) # ANOVA table
    
    p_value = anova_table["PR(>F)"][0]  # Get p-value
    print(f'{col}: p-value = {p_value:.4f}')
```
